```python
import numpy as np
from scipy.optimize import curve_fit

# =========================
# Modelo: doble lorentziana
# =========================
def doble_lorentziana(f, A1, f1, g1, A2, f2, g2, C):
    return (
        A1 / ((f - f1)**2 + (g1 / 2)**2) +
        A2 / ((f - f2)**2 + (g2 / 2)**2) +
        C
    )

# =========================
# Ajuste doble lorentziana
# =========================
def ajustar_doble_lorentziana(x, y, p0=None, bounds=None, maxfev=200000):
    """
    Ajusta una doble lorentziana a datos (x, y).

    Parámetros
    ----------
    x, y : array-like
        Datos experimentales.
    p0 : list o tuple, opcional
        Guess inicial [A1, f1, g1, A2, f2, g2, C].
        Si no se proporciona, se construye uno simple automáticamente.
    bounds : 2-tuple, opcional
        Cotas para curve_fit: (lower_bounds, upper_bounds).
    maxfev : int
        Número máximo de evaluaciones.

    Retorna
    -------
    resultado : dict
        Diccionario con:
        - popt: parámetros ajustados
        - pcov: matriz de covarianza
        - perr: incertidumbres 1-sigma
        - x_fit: eje suave para graficar
        - y_fit: curva ajustada
        - FWHM1, FWHM2
        - Q1, Q2
    """

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 7:
        raise ValueError("Se necesitan al menos 7 puntos para ajustar una doble lorentziana.")

    # Guess automático sencillo
    if p0 is None:
        idx_ord = np.argsort(y)[::-1]  # índices ordenados de mayor a menor
        idx1 = idx_ord[0]

        # buscar un segundo pico suficientemente separado
        idx2 = None
        separacion_min = max(3, len(x) // 20)
        for idx in idx_ord[1:]:
            if abs(idx - idx1) >= separacion_min:
                idx2 = idx
                break

        if idx2 is None:
            idx2 = idx_ord[1]

        # ordenar por frecuencia
        if x[idx1] < x[idx2]:
            i1, i2 = idx1, idx2
        else:
            i1, i2 = idx2, idx1

        C0 = np.min(y)
        A1_0 = max(y[i1] - C0, 1e-12) * ((x.max() - x.min())**2) / 10
        A2_0 = max(y[i2] - C0, 1e-12) * ((x.max() - x.min())**2) / 10
        f1_0 = x[i1]
        f2_0 = x[i2]
        g1_0 = max((x.max() - x.min()) / 30, 1e-6)
        g2_0 = max((x.max() - x.min()) / 30, 1e-6)

        p0 = [A1_0, f1_0, g1_0, A2_0, f2_0, g2_0, C0]

    # Bounds por defecto
    if bounds is None:
        lower = [0, x.min(), 1e-12, 0, x.min(), 1e-12, -np.inf]
        upper = [np.inf, x.max(), np.inf, np.inf, x.max(), np.inf, np.inf]
        bounds = (lower, upper)

    popt, pcov = curve_fit(
        doble_lorentziana,
        x, y,
        p0=p0,
        bounds=bounds,
        maxfev=maxfev
    )

    perr = np.sqrt(np.diag(pcov))

    # Reordenar para que el pico 1 tenga menor frecuencia que el pico 2
    if popt[1] > popt[4]:
        popt = np.array([popt[3], popt[4], popt[5], popt[0], popt[1], popt[2], popt[6]])
        perr = np.array([perr[3], perr[4], perr[5], perr[0], perr[1], perr[2], perr[6]])

    A1, f1, g1, A2, f2, g2, C = popt

    FWHM1 = abs(g1)
    FWHM2 = abs(g2)
    Q1 = f1 / FWHM1 if FWHM1 > 0 else np.inf
    Q2 = f2 / FWHM2 if FWHM2 > 0 else np.inf

    x_fit = np.linspace(x.min(), x.max(), 3000)
    y_fit = doble_lorentziana(x_fit, *popt)

    return {
        "popt": popt,
        "pcov": pcov,
        "perr": perr,
        "x_fit": x_fit,
        "y_fit": y_fit,
        "FWHM1": FWHM1,
        "FWHM2": FWHM2,
        "Q1": Q1,
        "Q2": Q2,
    }
```

```python
import os
import re
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# =========================
# Modelos
# =========================
def lorentziana(f, A0, f0, gamma, C):
    return A0 / ((f - f0)**2 + (gamma / 2)**2) + C

def doble_lorentziana(f, A1, f1, g1, A2, f2, g2, C):
    return (
        A1 / ((f - f1)**2 + (g1 / 2)**2) +
        A2 / ((f - f2)**2 + (g2 / 2)**2) +
        C
    )

# =========================
# Extraer modo n del nombre
# =========================
def extraer_modo(nombre_archivo):
    m = re.search(r"f(\d+)", nombre_archivo)
    if m:
        return int(m.group(1))
    return None

# =========================
# Ajuste simple
# =========================
def ajustar_lorentziana_simple(x, y, maxfev=200000):
    idx_max = np.argmax(y)
    f_res_exp = x[idx_max]
    A_max = y[idx_max]

    A0_guess = max(A_max - np.min(y), 1e-12) * ((x.max() - x.min())**2)
    f0_guess = f_res_exp
    gamma_guess = 0.5
    C_guess = np.min(y)

    p0 = [A0_guess, f0_guess, gamma_guess, C_guess]

    popt, pcov = curve_fit(
        lorentziana, x, y,
        p0=p0,
        maxfev=maxfev
    )

    A0_fit, f0_fit, gamma_fit, C_fit = popt
    FWHM = abs(gamma_fit)
    Q = f0_fit / FWHM if FWHM > 0 else np.inf

    f_fit = np.linspace(x.min(), x.max(), 2000)
    y_fit = lorentziana(f_fit, *popt)

    return {
        "tipo_ajuste": "simple",
        "popt": popt,
        "pcov": pcov,
        "f_fit": f_fit,
        "y_fit": y_fit,
        "f0": f0_fit,
        "FWHM": FWHM,
        "Q": Q
    }

# =========================
# Ajuste doble
# =========================
def ajustar_doble_lorentziana(x, y, p0=None, bounds=None, maxfev=200000):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if p0 is None:
        idx_ord = np.argsort(y)[::-1]
        idx1 = idx_ord[0]

        idx2 = None
        separacion_min = max(3, len(x) // 20)
        for idx in idx_ord[1:]:
            if abs(idx - idx1) >= separacion_min:
                idx2 = idx
                break

        if idx2 is None:
            idx2 = idx_ord[1]

        if x[idx1] < x[idx2]:
            i1, i2 = idx1, idx2
        else:
            i1, i2 = idx2, idx1

        C0 = np.min(y)
        ancho = x.max() - x.min()

        A1_0 = max(y[i1] - C0, 1e-12) * (ancho**2) / 10
        A2_0 = max(y[i2] - C0, 1e-12) * (ancho**2) / 10
        f1_0 = x[i1]
        f2_0 = x[i2]
        g1_0 = max(ancho / 30, 1e-6)
        g2_0 = max(ancho / 30, 1e-6)

        p0 = [A1_0, f1_0, g1_0, A2_0, f2_0, g2_0, C0]

    if bounds is None:
        lower = [0, x.min(), 1e-12, 0, x.min(), 1e-12, -np.inf]
        upper = [np.inf, x.max(), np.inf, np.inf, x.max(), np.inf, np.inf]
        bounds = (lower, upper)

    popt, pcov = curve_fit(
        doble_lorentziana, x, y,
        p0=p0,
        bounds=bounds,
        maxfev=maxfev
    )

    # Reordenar para que f1 < f2
    if popt[1] > popt[4]:
        popt = np.array([popt[3], popt[4], popt[5], popt[0], popt[1], popt[2], popt[6]])

    A1, f1, g1, A2, f2, g2, C = popt

    FWHM1 = abs(g1)
    FWHM2 = abs(g2)
    Q1 = f1 / FWHM1 if FWHM1 > 0 else np.inf
    Q2 = f2 / FWHM2 if FWHM2 > 0 else np.inf

    f_fit = np.linspace(x.min(), x.max(), 3000)
    y_fit = doble_lorentziana(f_fit, *popt)

    y_fit_1 = A1 / ((f_fit - f1)**2 + (g1 / 2)**2)
    y_fit_2 = A2 / ((f_fit - f2)**2 + (g2 / 2)**2)

    return {
        "tipo_ajuste": "doble",
        "popt": popt,
        "pcov": pcov,
        "f_fit": f_fit,
        "y_fit": y_fit,
        "y_fit_1": y_fit_1,
        "y_fit_2": y_fit_2,
        "f01": f1,
        "f02": f2,
        "FWHM1": FWHM1,
        "FWHM2": FWHM2,
        "Q1": Q1,
        "Q2": Q2
    }

# =========================
# Parámetros
# =========================
carpeta = "labequipo2/9abr_"

skip_idx = set()
skip_idx = {0,1,2,3,7,8}
xi,xf = [1654.5,2195.5,1188.0],[1655.2,2196.1,1189.75]
ext_ok = (".txt", ".dat", ".csv")
 
# Modos que quieres ajustar con DOBLE lorentziana
modos_dobles = {}
# ejemplo: modos_dobles = {2, 4}

# =========================
# Recolectar datos y ajustar
# =========================
archivos = sorted([fn for fn in os.listdir(carpeta) if fn.lower().endswith(ext_ok)])

resultados = []
fallos_ajuste = []
j = 0
for i, nombre_archivo in enumerate(archivos):
    if i in skip_idx:
        continue

    # if "Bueno" not in nombre_archivo:
    #     continue

    modo_n = extraer_modo(nombre_archivo)
    if modo_n is None:
        modo_n = 0
        # print(f"[WARN] No se pudo extraer el modo n de: {nombre_archivo}")
        # continue

    print(f"Procesando: {nombre_archivo} | modo n = {modo_n}")

    ruta = os.path.join(carpeta, nombre_archivo)

    try:
        a = np.loadtxt(ruta)
    except Exception as e:
        print(f"[WARN] No se pudo leer {nombre_archivo}: {e}")
        continue

    if a.ndim < 2 or a.shape[1] < 2:
        print(f"[WARN] Archivo inválido o con pocas columnas: {nombre_archivo}")
        continue

    x = a[:, 0].astype(float)
    y_db = a[:, 1].astype(float)

    # mask = (y_db > 0.0) & np.isfinite(y_db) & np.isfinite(x)
    ymax = np.max(y_db)
    # mask = (y_db > ymax - 20) & np.isfinite(y_db) & np.isfinite(x)
    mask = (x >= xi[j]) & (x <= xf[j])
    j += 1
    x = x[mask]
    y_db = y_db[mask]

    if len(x) < 5:
        print(f"[WARN] Muy pocos datos válidos en: {nombre_archivo}")
        continue

    y = 10 ** (y_db / 20.0)

    resultado = {
        "archivo": nombre_archivo,
        "modo_n": modo_n,
        "x": x,
        "y": y,
        "ajuste_ok": False,
        "tipo_ajuste": None
    }

    try:
        if modo_n in modos_dobles:
            ajuste = ajustar_doble_lorentziana(x, y)
        else:
            ajuste = ajustar_lorentziana_simple(x, y)

        resultado.update(ajuste)
        resultado["ajuste_ok"] = True

    except Exception as e:
        print(f"[WARN] Falló el ajuste para {nombre_archivo}: {e}")
        fallos_ajuste.append(nombre_archivo)

    resultados.append(resultado)

# =========================
# Ordenar por modo n
# =========================
resultados = sorted(resultados, key=lambda d: d["modo_n"])

n_paneles = len(resultados)
if n_paneles == 0:
    raise ValueError("No se encontraron archivos válidos para graficar.")

# =========================
# Estilo
# =========================
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 11,
    "legend.fontsize": 9,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 4,
    "ytick.major.size": 4,
})

# =========================
# Subplots dinámicos
# =========================
ncols = 2
nrows = math.ceil(n_paneles / ncols)

fig, axes = plt.subplots(nrows, ncols, figsize=(10.5, 3.2 * nrows), dpi=300)
axes = np.atleast_1d(axes).ravel()

# =========================
# Graficar
# =========================
for k, d in enumerate(resultados):
    ax = axes[k]

    ax.plot(
        d["x"], d["y"],
        'o',
        markersize=3.8,
        markerfacecolor='white',
        markeredgewidth=0.9,
        label="Datos"
    )

    titulo = rf"Modo $\beta_{d['modo_n']+3}$"

    if d["ajuste_ok"]:
        if d["tipo_ajuste"] == "simple":
            ax.plot(
                d["f_fit"], d["y_fit"],
                '-', linewidth=1.2,
                label="Ajuste lorentziano"
            )

            ax.axvline(d["f0"], linestyle="--", linewidth=1.0)

            info = "\n".join([
                "Ajuste simple",
                rf"$f_0 = {d['f0']:.4f}\ \mathrm{{Hz}}$",
                rf"$\mathrm{{FWHM}} = {d['FWHM']:.4f}\ \mathrm{{Hz}}$",
                rf"$Q = {d['Q']:.0f}$"
            ])

        elif d["tipo_ajuste"] == "doble":
            ax.plot(
                d["f_fit"], d["y_fit"],
                '-', linewidth=1.2,
                label="Ajuste doble"
            )
            # ax.plot(
            #     d["f_fit"], d["y_fit_1"],
            #     '--', linewidth=1.0,
            #     label="Pico 1"
            # )
            # ax.plot(
            #     d["f_fit"], d["y_fit_2"],
            #     '--', linewidth=1.0,
            #     label="Pico 2"
            # )

            ax.axvline(d["f01"], linestyle="--", linewidth=1.0)
            ax.axvline(d["f02"], linestyle="--", linewidth=1.0)

            info = "\n".join([
                "Ajuste doble",
                rf"$f_1 = {d['f01']:.4f}\ \mathrm{{Hz}}$",
                rf"$f_2 = {d['f02']:.4f}\ \mathrm{{Hz}}$",
                rf"$Q_1 = {d['Q1']:.0f}$",
                rf"$Q_2 = {d['Q2']:.0f}$"
            ])
    else:
        info = "Ajuste no convergió"

    ax.set_title(titulo)
    ax.set_xlabel(r"Frecuencia $f$ (Hz)")
    ax.set_ylabel(r"Amplitud (u.a.)")

    ax.text(
        0.03, 0.97, info,
        transform=ax.transAxes,
        va="top", ha="left",
        bbox=dict(boxstyle="round", facecolor="white", edgecolor="black", alpha=0.95)
    )

    ax.legend(frameon=False, loc="upper right")
    ax.grid(True, alpha=0.25)

# Apagar paneles sobrantes
for j in range(n_paneles, len(axes)):
    axes[j].axis("off")

fig.tight_layout()
plt.rcParams["axes.labelsize"] = 9
plt.show()

# =========================
# Resumen en consola
# =========================
print("\n===== RESUMEN =====")
for d in resultados:
    if not d["ajuste_ok"]:
        print(f"n={d['modo_n']} | {d['archivo']} | AJUSTE FALLIDO")
        continue

    if d["tipo_ajuste"] == "simple":
        print(
            f"n={d['modo_n']} | {d['archivo']} | simple | "
            f"f0={d['f0']:.4f} Hz | FWHM={d['FWHM']:.4f} Hz | Q={d['Q']:.2f}"
        )
    else:
        print(
            f"n={d['modo_n']} | {d['archivo']} | doble | "
            f"f1={d['f01']:.4f} Hz | FWHM1={d['FWHM1']:.4f} Hz | Q1={d['Q1']:.2f} | "
            f"f2={d['f02']:.4f} Hz | FWHM2={d['FWHM2']:.4f} Hz | Q2={d['Q2']:.2f}"
        )

if fallos_ajuste:
    print("\n===== AJUSTES QUE FALLARON =====")
    for nombre in fallos_ajuste:
        print(f"- {nombre}")
```

```python
import numpy as np
from math import pi

# ====== Parámetros (SI) ======
a = 9.98e-3
b = 19.36e-3
L = 0.7510
rho = 2551.294
E = 66.9709e9  # Pa

A = a*b
Iy = (b*a**3)/12
Iz = (a*b**3)/12

# Raíces beta_n (libre–libre, no rígidos) suficientes para cubrir varios modos
beta = np.array([
    4.730040744862705,
    7.853204624095838,
    10.995607838001671,
    14.137165491257,
    17.278759657399,
    20.420352245626,
    23.561944902040,
    26.703537555510,
    29.845130209103,
    32.986722862692,
], float)

def flex_freq(beta, I):
    return (beta**2/(2*pi*L**2))*np.sqrt(E*I/(rho*A))

fy = flex_freq(beta, Iy)
fz = flex_freq(beta, Iz)

# ====== Tus frecuencias experimentales (de la figura) ======
f_exp = np.array([1188.8088, 1654.8509, 2195.8360], float)

def match_family(f_exp, f_theo, beta):
    # devuelve: (idx_mode, beta_n, f_theo, err_rel)
    out = []
    for fe in f_exp:
        j = np.argmin(np.abs(f_theo - fe))
        err = (fe - f_theo[j]) / f_theo[j]
        out.append((j+1, beta[j], f_theo[j], err))
    return out

m_y = match_family(f_exp, fy, beta)
m_z = match_family(f_exp, fz, beta)

print("=== Matching con familia Iy ===")
for fe, (n, bn, ft, err) in zip(f_exp, m_y):
    print(f"f_exp={fe:9.3f} Hz  ->  n={n:2d}, beta={bn:7.4f}, f_theo={ft:9.3f} Hz,  err={100*err:+6.2f}%")

print("\n=== Matching con familia Iz ===")
for fe, (n, bn, ft, err) in zip(f_exp, m_z):
    print(f"f_exp={fe:9.3f} Hz  ->  n={n:2d}, beta={bn:7.4f}, f_theo={ft:9.3f} Hz,  err={100*err:+6.2f}%")

# Decisión simple: promedio del |error| en cada familia
avg_err_y = np.mean([abs(e) for *_, e in m_y])
avg_err_z = np.mean([abs(e) for *_, e in m_z])
print("\nPromedio |error|:  Iy =", 100*avg_err_y, "%   Iz =", 100*avg_err_z, "%")
```
