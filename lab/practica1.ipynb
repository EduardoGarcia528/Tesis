```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Datos experimentales
# -----------------------------
n = np.array([1, 2, 3, 4, 5], dtype=float)

f = np.array([
    3415.0344,
    6830.6349,
    10243.0020,
    13651.9560,
    17059.8419
], dtype=float)

L = 0.75  # m

# -----------------------------
# Ajuste lineal
# -----------------------------
m, b = np.polyfit(n, f, 1)
f_fit = m * n + b
v_exp = 2 * L * m

# Coeficiente de determinación R^2
ss_res = np.sum((f - f_fit)**2)
ss_tot = np.sum((f - np.mean(f))**2)
R2 = 1 - ss_res / ss_tot

# Curva suave para la recta
n_smooth = np.linspace(0.8, 5.2, 300)
f_smooth = m * n_smooth + b

# -----------------------------
# Estilo tipo paper
# -----------------------------
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 5,
    "ytick.major.size": 5,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
})

fig, ax = plt.subplots(figsize=(6.2, 4.6), dpi=300)

# Datos experimentales
ax.plot(
    n, f,
    'o',
    markersize=6,
    markerfacecolor='white',
    markeredgewidth=1.2,
    label='Datos experimentales'
)

# Ajuste lineal
ax.plot(
    n_smooth, f_smooth,
    linewidth=1.5,
    label='Ajuste lineal'
)

# Etiquetas
ax.set_xlabel(r'Modo normal, $n$')
ax.set_ylabel(r'Frecuencia resonante, $f_n$ (Hz)')

# Límites y ticks
ax.set_xlim(0.7, 5.3)
ax.set_xticks([1, 2, 3, 4, 5])

# Texto con resultados
textstr = '\n'.join((
    rf'$f_n = ({m:.2f})\,n + ({b:.2f})$',
    rf'$v = 2Lm = {v_exp:.2f}\ \mathrm{{m/s}}$',
    rf'$R^2 = {R2:.8f}$'
))

ax.text(
    0.05, 0.95, textstr,
    transform=ax.transAxes,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='white', edgecolor='black', alpha=1.0)
)

# Leyenda
ax.legend(frameon=False, loc='lower right')

# Ajuste fino del marco
for spine in ax.spines.values():
    spine.set_linewidth(1.0)

fig.tight_layout()

# Guardar figura en alta resolución
plt.savefig('ajuste_lineal_montaje2.png', dpi=300, bbox_inches='tight')
plt.savefig('ajuste_lineal_montaje2.pdf', bbox_inches='tight')

plt.show()

# -----------------------------
# Imprimir resultados
# -----------------------------
print("===== AJUSTE LINEAL =====")
print(f"Pendiente m = {m:.4f} Hz")
print(f"Intercepto b = {b:.4f} Hz")
print(f"Velocidad experimental v = {v_exp:.4f} m/s")
print(f"R^2 = {R2:.8f}")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# =========================
# Modelo Lorentziano
# =========================
def lorentziana(f, A0, f0, gamma, C):
    return A0 / ((f - f0)**2 + (gamma/2)**2) + C

# =========================
# Parámetros
# =========================
carpeta = "labequipo2"

# Si quieres excluir índices específicos como antes, edítalo aquí:
skip_idx = {0, 2, 3, 5, 6, 9, 11, 12, 13, 14, 15}

# Extensiones permitidas
ext_ok = (".txt", ".dat", ".csv")

# =========================
# Recolectar y ajustar (primero)
# =========================
archivos = sorted([fn for fn in os.listdir(carpeta) if fn.lower().endswith(ext_ok)])

resultados = []
for i, nombre_archivo in enumerate(archivos):
    if i in skip_idx:
        continue

    ruta = os.path.join(carpeta, nombre_archivo)
    a = np.loadtxt(ruta)

    x = a[:, 0].astype(float)
    y_db = a[:, 1].astype(float)

    # Filtrar ceros (solo en dB)
    mask = (y_db != 0.0) & np.isfinite(y_db) & np.isfinite(x)
    x = x[mask]
    y_db = y_db[mask]

    # Convertir a amplitud lineal
    y = 10**(y_db / 20.0)

    # Pico inicial
    idx_max = np.argmax(y)
    f_res_exp = x[idx_max]
    A_max = y[idx_max]

    # Guess inicial (robusto)
    A0_guess = max(A_max - np.min(y), 1e-12) * ((x.max() - x.min())**2)
    f0_guess = f_res_exp
    gamma_guess = 0.5  # Hz (tu caso tiene picos angostos)
    C_guess = np.min(y)

    p0 = [A0_guess, f0_guess, gamma_guess, C_guess]

    # Ajuste
    try:
        popt, pcov = curve_fit(
            lorentziana, x, y, p0=p0,
            maxfev=200000
        )
    except RuntimeError:
        print(f"[WARN] No convergió el ajuste para: {nombre_archivo}")
        continue

    A0_fit, f0_fit, gamma_fit, C_fit = popt
    FWHM = abs(gamma_fit)
    Q = f0_fit / FWHM if FWHM > 0 else np.inf

    # Curva suave
    f_fit = np.linspace(x.min(), x.max(), 2000)
    y_fit = lorentziana(f_fit, *popt)

    resultados.append({
        "archivo": nombre_archivo,
        "x": x, "y": y,
        "f_fit": f_fit, "y_fit": y_fit,
        "f0": f0_fit, "FWHM": FWHM, "Q": Q
    })

# Ordenar por frecuencia (para que queden n=1..5)
resultados = sorted(resultados, key=lambda d: d["f0"])

# Tomar exactamente 5 (si hay más)
resultados = resultados[:5]

if len(resultados) < 5:
    raise ValueError(f"Se encontraron {len(resultados)} resonancias ajustadas; se esperaban 5.")

# =========================
# Estilo tipo paper
# =========================
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 11,
    "legend.fontsize": 9,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 4,
    "ytick.major.size": 4,
})

# =========================
# Subplots 3x2 (5 gráficas + 1 vacío)
# =========================
fig, axes = plt.subplots(3, 2, figsize=(10.5, 9.0), dpi=300)
axes = axes.ravel()

for k, d in enumerate(resultados):
    ax = axes[k]

    # Datos
    ax.plot(
        d["x"], d["y"],
        'o', markersize=3.8,
        markerfacecolor='white',
        markeredgewidth=0.9,
    )

    # Ajuste
    ax.plot(
        d["f_fit"], d["y_fit"],
        '-', linewidth=1.2,
    )

    # Línea vertical en f0
    ax.axvline(d["f0"], linestyle="--", linewidth=1.0)

    # Título tipo "n = k+1"
    ax.set_title(rf"Modo $n={k+1}$")

    ax.set_xlabel(r"Frecuencia $f$ (Hz)")
    ax.set_ylabel(r"Amplitud (u.a.)")

    # Anotación con f0, FWHM, Q
    info = "\n".join([
        rf"$f_0 = {d['f0']:.4f}\ \mathrm{{Hz}}$",
        rf"$\mathrm{{FWHM}} = {d['FWHM']:.4f}\ \mathrm{{Hz}}$",
        rf"$Q = {d['Q']:.0f}$"
    ])

    ax.text(
        0.03, 0.97, info,
        transform=ax.transAxes,
        va="top", ha="left",
        bbox=dict(boxstyle="round", facecolor="white", edgecolor="black", alpha=0.95)
    )

    # Leyenda discreta
    ax.legend(frameon=False, loc="lower right")

    # Grid sutil (opcional: si no quieres, comenta esta línea)
    ax.grid(True, alpha=0.25)

# Apagar el sexto panel (vacío)
axes[-1].axis("off")

# fig.suptitle("Montaje 2 (L = 0.75 m): Resonancias y ajustes Lorentzianos", y=0.995)
fig.tight_layout()

# Guardar figura
# plt.savefig("montaje2_subplots_lorentz.png", dpi=300, bbox_inches="tight")
# plt.savefig("montaje2_subplots_lorentz.pdf", bbox_inches="tight")

plt.show()

# =========================
# Resumen en consola
# =========================
print("\n===== RESUMEN (ordenado por f0) =====")
for k, d in enumerate(resultados, start=1):
    print(f"n={k} | {d['archivo']} | f0={d['f0']:.4f} Hz | FWHM={d['FWHM']:.4f} Hz | Q={d['Q']:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Datos experimentales
# =========================
n = np.array([1, 2, 3, 4, 5], dtype=float)

f = np.array([
    3415.0344,
    6830.6349,
    10243.0020,
    13651.9560,
    17059.8419
], dtype=float)

L = 0.75  # longitud de la varilla en metros

# =========================
# Ajuste lineal
# =========================
coef = np.polyfit(n, f, 1)   # coef[0]=pendiente, coef[1]=intercepto
m, b = coef

f_fit = m*n + b

# Velocidad experimental
v_exp = 2 * L * m

# R^2
ss_res = np.sum((f - f_fit)**2)
ss_tot = np.sum((f - np.mean(f))**2)
R2 = 1 - ss_res/ss_tot

print("===== AJUSTE LINEAL =====")
print(f"Pendiente m = {m:.4f} Hz")
print(f"Intercepto b = {b:.4f} Hz")
print(f"Velocidad experimental v = {v_exp:.4f} m/s")
print(f"R^2 = {R2:.8f}")

# =========================
# Gráfica
# =========================
n_suave = np.linspace(1, 5, 200)
f_suave = m*n_suave + b

plt.figure(figsize=(8,5))
plt.plot(n, f, '.', markersize=8, label='Datos experimentales')
plt.plot(n_suave, f_suave, '-', label='Ajuste lineal')

# Etiquetas de cada punto
for ni, fi in zip(n, f):
    plt.text(ni + 0.03, fi + 20, f'({int(ni)}, {fi:.1f})', fontsize=9)

plt.xlabel('Número de modo n')
plt.ylabel('Frecuencia resonante $f_n$ (Hz)')
plt.title('Montaje 2: ajuste lineal de $f_n$ vs $n$')
plt.grid(True)
plt.legend()

texto = (
    f'$f_n = ({m:.2f})n + ({b:.2f})$\n'
    f'$v_{{exp}} = {v_exp:.2f}\\,\\mathrm{{m/s}}$\n'
    f'$R^2 = {R2:.8f}$'
)

plt.text(0.05, 0.95, texto,
         transform=plt.gca().transAxes,
         verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='white', alpha=0.85))

plt.tight_layout()
plt.show()
```
