```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Datos experimentales
# -----------------------------
n = np.array([1, 2, 3, 4, 5], dtype=float)

f = np.array([
    3415.0344,
    6830.6349,
    10243.0020,
    13651.9560,
    17059.8419
], dtype=float)

L = 0.75  # m

# -----------------------------
# Ajuste lineal
# -----------------------------
m, b = np.polyfit(n, f, 1)
f_fit = m * n + b
v_exp = 2 * L * m

# Coeficiente de determinación R^2
ss_res = np.sum((f - f_fit)**2)
ss_tot = np.sum((f - np.mean(f))**2)
R2 = 1 - ss_res / ss_tot

# Curva suave para la recta
n_smooth = np.linspace(0.8, 5.2, 300)
f_smooth = m * n_smooth + b

# -----------------------------
# Estilo tipo paper
# -----------------------------
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 5,
    "ytick.major.size": 5,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
})

fig, ax = plt.subplots(figsize=(6.2, 4.6), dpi=300)

# Datos experimentales
ax.plot(
    n, f,
    'o',
    markersize=6,
    markerfacecolor='white',
    markeredgewidth=1.2,
    label='Datos experimentales'
)

# Ajuste lineal
ax.plot(
    n_smooth, f_smooth,
    linewidth=1.5,
    label='Ajuste lineal'
)

# Etiquetas
ax.set_xlabel(r'Modo normal, $n$')
ax.set_ylabel(r'Frecuencia resonante, $f_n$ (Hz)')

# Límites y ticks
ax.set_xlim(0.7, 5.3)
ax.set_xticks([1, 2, 3, 4, 5])

# Texto con resultados
textstr = '\n'.join((
    rf'$f_n = ({m:.2f})\,n + ({b:.2f})$',
    rf'$v = 2Lm = {v_exp:.2f}\ \mathrm{{m/s}}$',
    rf'$R^2 = {R2:.8f}$'
))

ax.text(
    0.05, 0.95, textstr,
    transform=ax.transAxes,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='white', edgecolor='black', alpha=1.0)
)

# Leyenda
ax.legend(frameon=False, loc='lower right')

# Ajuste fino del marco
for spine in ax.spines.values():
    spine.set_linewidth(1.0)

fig.tight_layout()

# Guardar figura en alta resolución
plt.savefig('ajuste_lineal_montaje2.png', dpi=300, bbox_inches='tight')
plt.savefig('ajuste_lineal_montaje2.pdf', bbox_inches='tight')

plt.show()

# -----------------------------
# Imprimir resultados
# -----------------------------
print("===== AJUSTE LINEAL =====")
print(f"Pendiente m = {m:.4f} Hz")
print(f"Intercepto b = {b:.4f} Hz")
print(f"Velocidad experimental v = {v_exp:.4f} m/s")
print(f"R^2 = {R2:.8f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


# =========================
# 2. CONVERSIÓN dB -> AMPLITUD
# =========================
import os

carpeta = "labequipo2"

for nombre_archivo in os.listdir(carpeta):
    print(nombre_archivo)
    ruta_completa = os.path.join(carpeta, nombre_archivo)
    a = np.loadtxt(ruta_completa)
    # x = a[:,0]
    x = a[:,0]
    y_db = a[:,1]
    # y_db = a[:,1] 
    y = 10**(y_db/20)

    # =========================
    # 3. DETECCIÓN DEL PICO (frecuencia de resonancia experimental)
    # =========================
    idx_max = np.argmax(y)
    f_res_exp = x[idx_max]
    A_max = y[idx_max]

    # print("===== RESULTADOS BÁSICOS =====")
    print(f"Frecuencia de resonancia (máximo): {f_res_exp:.4f} Hz")
    # print(f"Amplitud máxima (lineal): {A_max:.6e}")

    # =========================
    # 4. FUNCIÓN LORENTZIANA PARA AJUSTE
    # =========================
    def lorentziana(f, A0, f0, gamma, C):
        return A0 / ((f - f0)**2 + (gamma/2)**2) + C

    # Valores iniciales para el ajuste
    A0_guess = A_max * ( (x[-1] - x[0])**2 )
    f0_guess = f_res_exp
    gamma_guess = 1.0
    C_guess = np.min(y)

    p0 = [A0_guess, f0_guess, gamma_guess, C_guess]

    # Ajuste no lineal
    popt, pcov = curve_fit(lorentziana, x, y, p0=p0, maxfev=100000)
    A0_fit, f0_fit, gamma_fit, C_fit = popt

    # =========================
    # 5. FWHM Y FACTOR DE CALIDAD Q
    # =========================
    # Para una Lorentziana: FWHM = gamma
    FWHM = gamma_fit
    Q = f0_fit / FWHM

    print("\n===== AJUSTE LORENTZIANO =====")
    print(f"Frecuencia de resonancia ajustada f0: {f0_fit:.4f} Hz")
    print(f"Ancho de resonancia FWHM (gamma): {FWHM:.4f} Hz")
    print(f"Factor de calidad Q: {Q:.2f}")

    # =========================
    # 6. COMPARACIÓN CON TEORÍA (varilla libre-libre, L = 1 m)
    # =========================
    L = 0.75  # metros
    E = 70e9  # Pa (Aluminio)
    rho = 2700  # kg/m^3

    v_teo = np.sqrt(E / rho)
    f1_teo = v_teo / (2 * L)

    error_rel = abs(f0_fit - f1_teo) / f1_teo * 100

    # print("\n===== COMPARACIÓN TEÓRICA =====")
    # print(f"Velocidad teórica del sonido en Al: {v_teo:.2f} m/s")
    # print(f"Frecuencia teórica modo n=1: {f1_teo:.2f} Hz")
    # print(f"Error relativo: {error_rel:.2f} %")

    # =========================
    # 7. GRÁFICAS (estándar de reporte)
    # =========================
    f_fit = np.linspace(x.min(), x.max(), 2000)
    y_fit = lorentziana(f_fit, *popt)

    plt.figure(figsize=(8,5))
    plt.plot(x, y, 'o', label='Datos experimentales')
    plt.plot(f_fit, y_fit, '-', label='Ajuste Lorentziano')
    plt.axvline(f0_fit, linestyle='--', label=f'f1 = {f0_fit:.2f} Hz')
    plt.xlabel('Frecuencia (Hz)')
    plt.ylabel('Amplitud (escala lineal)')
    plt.title('Resonancia compresional - Varilla de aluminio (L = 0.75 m)')
    plt.legend()
    plt.grid(True)
    plt.show()

    # =========================
    # 8. GRÁFICA EN dB (útil para el reporte)
    # =========================
    plt.figure(figsize=(8,5))
    plt.plot(x, y_db)
    plt.axvline(f0_fit, linestyle='--')
    plt.xlabel('Frecuencia (Hz)')
    plt.ylabel('Amplitud (dB)')
    plt.title('Espectro de resonancia en dB')
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Datos experimentales
# =========================
n = np.array([1, 2, 3, 4, 5], dtype=float)

f = np.array([
    3415.0344,
    6830.6349,
    10243.0020,
    13651.9560,
    17059.8419
], dtype=float)

L = 0.75  # longitud de la varilla en metros

# =========================
# Ajuste lineal
# =========================
coef = np.polyfit(n, f, 1)   # coef[0]=pendiente, coef[1]=intercepto
m, b = coef

f_fit = m*n + b

# Velocidad experimental
v_exp = 2 * L * m

# R^2
ss_res = np.sum((f - f_fit)**2)
ss_tot = np.sum((f - np.mean(f))**2)
R2 = 1 - ss_res/ss_tot

print("===== AJUSTE LINEAL =====")
print(f"Pendiente m = {m:.4f} Hz")
print(f"Intercepto b = {b:.4f} Hz")
print(f"Velocidad experimental v = {v_exp:.4f} m/s")
print(f"R^2 = {R2:.8f}")

# =========================
# Gráfica
# =========================
n_suave = np.linspace(1, 5, 200)
f_suave = m*n_suave + b

plt.figure(figsize=(8,5))
plt.plot(n, f, '.', markersize=8, label='Datos experimentales')
plt.plot(n_suave, f_suave, '-', label='Ajuste lineal')

# Etiquetas de cada punto
for ni, fi in zip(n, f):
    plt.text(ni + 0.03, fi + 20, f'({int(ni)}, {fi:.1f})', fontsize=9)

plt.xlabel('Número de modo n')
plt.ylabel('Frecuencia resonante $f_n$ (Hz)')
plt.title('Montaje 2: ajuste lineal de $f_n$ vs $n$')
plt.grid(True)
plt.legend()

texto = (
    f'$f_n = ({m:.2f})n + ({b:.2f})$\n'
    f'$v_{{exp}} = {v_exp:.2f}\\,\\mathrm{{m/s}}$\n'
    f'$R^2 = {R2:.8f}$'
)

plt.text(0.05, 0.95, texto,
         transform=plt.gca().transAxes,
         verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='white', alpha=0.85))

plt.tight_layout()
plt.show()
```
