```python
import os
import numpy as np
import matplotlib.pyplot as plt

folder_path = 'new_data/xi_index1'
# folder_path = 'data/xi_index1'
composer = []
plt.figure(figsize=(10, 6))
for filename in os.listdir(folder_path):
    composer.append(filename[5:-7])
    if filename.endswith('.npy'):
        array = np.load(os.path.join(folder_path, filename))[:,1]
        plt.hist(array, bins='auto', alpha=0.5, label=filename[5:-7], edgecolor='black')
print(composer)
# plt.title('Histograma de los datos de cada array en xi_index')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```
