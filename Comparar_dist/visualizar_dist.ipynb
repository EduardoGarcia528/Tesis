```python
import os
import numpy as np
import matplotlib.pyplot as plt

folder_path = 'xi_index_mdfa'
folder_path = 'Js'

plt.figure(figsize=(10, 6))
for filename in os.listdir(folder_path):
    if filename.endswith('.npy'):
        print(filename)
        array = np.load(os.path.join(folder_path, filename))
        print(array)
        plt.hist(array, bins='auto', alpha=0.5, label=filename[5:-7], edgecolor='black')

plt.title('Histograma de los datos de cada array en xi_index')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```
