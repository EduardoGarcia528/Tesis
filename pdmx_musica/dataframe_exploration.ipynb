```python
import pandas as pd
pd.set_option("display.max_rows", None)
df = pd.read_csv('PDMX/PDMX.csv')

print("\n=== Información de columnas ===")
info_cols = pd.DataFrame({
    "columna": df.columns,
    "dtype": df.dtypes.values,
    "n_nans": df.isna().sum().values
})
print(info_cols)
```

```python
"""Artistas por género"""
for i in df.loc[df["genres"] == 'classical', "artist_name"].unique():
    print(i)
```

```python
"""Canciones del artista"""
df_filtrado = df[df["artist_name"].str.contains("Paganini", na=False)]
for i in df_filtrado['song_name']:
    print(i)
```

```python
palabras = ["sonata", "violin"]

mask = df["song_name"].fillna("").str.lower().apply(
    lambda x: all(p.lower() in x for p in palabras)
)

resultados = df[mask]
for nombre, path, artista in zip(resultados["song_name"], resultados["path"], resultados['artist_name']):
    print(nombre,"|" ,artista, "->", path)
```
